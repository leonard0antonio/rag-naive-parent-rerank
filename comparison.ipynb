# Comparação dos 3 Modelos RAG - Os Sertões

Este notebook carrega os resultados dos 3 modelos (Naive, Parent, Rerank) e faz uma análise comparativa.

```python
# Carregar resultados
import json

with open("naive_rag_results.json", "r", encoding="utf-8") as f:
    naive_results = json.load(f)

with open("parent_rag_results.json", "r", encoding="utf-8") as f:
    parent_results = json.load(f)

with open("rerank_rag_results.json", "r", encoding="utf-8") as f:
    rerank_results = json.load(f)

print("Resultados carregados com sucesso!")
```

```python
# Comparação lado a lado
questions = [
    "Visão sobre ambiente natural do sertão",
    "Características da população sertaneja",
    "Contexto histórico/político Guerra de Canudos",
    "Figura de Antônio Conselheiro",
    "Crítica social e política em Os Sertões"
]
```

```python
for i, q_short in enumerate(questions, 1):
    key = f"q{i}"
    print(f"\n{'='*100}")
    print(f"PERGUNTA {i}: {q_short}")
    print(f"{'='*100}")
    
    print(f"\n--- NAIVE RAG ---")
    print(naive_results[key]["answer"][:500] + "...")
    
    print(f"\n--- PARENT RAG ---")
    print(parent_results[key]["answer"][:500] + "...")
    
    print(f"\n--- RERANK RAG ---")
    print(rerank_results[key]["answer"][:500] + "...")
```

```python
# Análise de fontes utilizadas
print("ANÁLISE DE FONTES UTILIZADAS")
print("="*80)

for i in range(1, 6):
    key = f"q{i}"
    print(f"\nPergunta {i}:")
    
    naive_sources = len(naive_results[key]["sources"])
    parent_sources = len(parent_results[key]["sources"])
    rerank_sources = len(rerank_results[key]["sources"])
    
    print(f"  Naive:  {naive_sources} chunks")
    print(f"  Parent: {parent_sources} páginas completas")
    print(f"  Rerank: {rerank_sources} chunks rerankeados")
    
    # Scores médios
    if naive_results[key]["sources"]:
        avg_naive = sum(s["score"] for s in naive_results[key]["sources"]) / naive_sources
        print(f"  Naive avg score: {avg_naive:.4f}")
    
    if parent_results[key]["sources"]:
        avg_parent = sum(s["score"] for s in parent_results[key]["sources"]) / parent_sources
        print(f"  Parent avg score: {avg_parent:.4f}")
    
    if rerank_results[key]["sources"]:
        avg_rerank = sum(s["rerank_score"] for s in rerank_results[key]["sources"]) / rerank_sources
        avg_bi = sum(s["bi_encoder_score"] for s in rerank_results[key]["sources"]) / rerank_sources
        print(f"  Rerank avg score: {avg_rerank:.4f} (bi-encoder: {avg_bi:.4f})")
```

## Resumo Comparativo

### Quando usar cada abordagem:

| Cenário | Recomendação |
|---------|--------------|
| Prototipagem rápida, baseline | **Naive RAG** |
| Documentos longos (livros, papers, legais) | **Parent RAG** |
| Produção, alta precisão necessária | **Rerank RAG** |
| Perguntas factuais simples | Naive RAG suficiente |
| Perguntas que precisam contexto amplo | Parent RAG |
| Perguntas complexas, ambíguas, sutis | Rerank RAG |

### Trade-offs:

- **Naive RAG**: Mais rápido, mais simples, mas pode perder contexto
- **Parent RAG**: Melhor contexto, boa para documentos estruturados, busca menos precisa
- **Rerank RAG**: Mais preciso semanticamente, mas 2-3x mais lento (cross-encoder)

### Para "Os Sertões" especificamente:
- O livro é longo (~500 páginas) e denso
- **Parent RAG** aproveita a estrutura por páginas
- **Rerank RAG** é melhor para perguntas analíticas (crítica social, visão do autor)
- **Naive RAG** serve como baseline para medir melhorias
